# Car Price Prediction - Multiple Linear Regression
Implemented from scratch with NumPy and gradient descent (no scikit-learn), using only the provided dataset.

```python
import numpy as np
import pandas as pd
```

## Load the data
Only the numeric columns of the dataset are used. `car_ID` is just an identifier so it is dropped; `price` is the target.

```python
df = pd.read_csv('CarPrice_Assignment.csv')
df = df.drop(columns=['car_ID'])
df = df.select_dtypes(include=np.number)

targets = df['price'].values
features = df.drop(columns='price').values.astype(float)
print(features.shape)
```

## Normalization
`features` is a 2-D array: each **row** is a car and each **column** is a feature. `features.T` is the transpose (rows and columns swapped), so looping over `features.T` gives one feature (column) at a time. Each feature is mean-normalized: (x - mean) / (max - min).

```python
def normalize(features):
    for feature in features.T:
        fmean = np.mean(feature)
        frange = np.amax(feature) - np.amin(feature)
        feature -= fmean
        feature /= frange
    return features
```

## Prediction, cost function and gradient descent

```python
def predict(features, weights):
    return np.dot(features, weights)

def cost_function(features, targets, weights):
    N = len(targets)
    predictions = predict(features, weights)
    sq_error = (predictions - targets)**2
    return 1.0/(2*N) * sq_error.sum()

def update_weights(features, targets, weights, lr):
    predictions = predict(features, weights)
    gradient = np.dot(-features.T, targets - predictions)
    gradient /= len(features)
    weights -= lr * gradient
    return weights
```

## Training

```python
def train(features, targets, weights, lr, iters):
    cost_history = []
    for i in range(iters):
        weights = update_weights(features, targets, weights, lr)
        cost = cost_function(features, targets, weights)
        cost_history.append(cost)
        if i % 500 == 0:
            print("iter={:d}    cost={:.2f}".format(i, cost))
    return weights, cost_history
```

```python
features = normalize(features)

# Bias term: column of ones added as the first column
bias = np.ones(shape=(len(features), 1))
features = np.append(bias, features, axis=1)

weights = np.zeros(features.shape[1])
weights, cost_history = train(features, targets, weights, 0.1, 5000)
```

## Result

```python
print("final cost =", cost_function(features, targets, weights))
print("weights =", weights)
```
